This notebook does not rerun family-level feature validation. It assembles the final family packages, verifies that the expected inputs are present, harmonizes feature roles, and builds a concise support figure for Results Section 3.1.

The figure is intentionally limited to three messages:

1. representative measurements respond as expected under controlled perturbation;
2. principal indicators show empirical persistence across repeated recordings where estimable; and
3. validation assigned different analytical roles rather than promoting every candidate measurement.

## Interpretation boundary

Panel B shows empirical persistence between the first two available recordings from a participant. These recordings were obtained under changing acquisition and disease conditions and are not technical replicate recordings. The estimates should therefore be described as repeated-recording persistence rather than technical test–retest reliability.

QDIST positive-part persistence was not estimable because very few participants were positive at both recordings. QTEMP is displayed as a completed negative feature-selection result and contributes no primary indicator.
